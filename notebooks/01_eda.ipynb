```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../backend/ml/data/raw/mumbai.csv", parse_dates=["time"], index_col="time")

# 1. Plot temperature over time
df["temperature_2m"].plot(figsize=(15, 4), title="Mumbai Temperature")
plt.show()

# 2. Zoom into one week
df["temperature_2m"]["2023-05-01":"2023-05-08"].plot(figsize=(15,4), title="One week")
plt.show()

# 3. Check for missing values
print(df.isna().sum())

# 4. Stats
print(df.describe())
```
